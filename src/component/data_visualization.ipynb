```python
import matplotlib.pyplot as plt
import pandas as pd 
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('/Users/ankitbista/Desktop/practice/HousePricePrediction/dataset.csv')
df.head(5)
```

```python
sns.scatterplot(x=df['Price'],y=df['Area'])
plt.show()
```

```python
correlation_matrix = df.corr()

plt.figure(figsize=(10,8))
sns.heatmap(correlation_matrix,annot=True,cmap='coolwarm',linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.boxplot(data=df[['Price', 'Area' , 'HouseAge','Bedrooms','Bathrooms']])
plt.title("Boxplot of Numerical Features")
```

```python
plt.figure(figsize=(10, 6))
df['Price'].hist(bins=30, color='skyblue', edgecolor='black')
plt.title('Distribution of Price')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()

```

```python
plt.figure(figsize=(10, 6))
sns.lineplot(x='YearBuilt', y='Price', data=df, marker='o')
plt.title('Price over Year Built')
plt.xlabel('Year Built')
plt.ylabel('Price')
plt.show()
```
