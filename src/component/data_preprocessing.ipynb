```python
import numpy as np 
import pickle
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('/Users/ankitbista/Desktop/practice/HousePricePrediction/dataset.csv')
```

```python
df.head(5)
```

```python
encoder = LabelEncoder()
df['Location'] = encoder.fit_transform(df['Location'])
df
```

```python
df['Condition'] = encoder.fit_transform(df['Condition'])
df
```

```python
df['Garage'] = encoder.fit_transform(df['Garage'])
df
```

```python
df.to_csv('/Users/ankitbista/Desktop/practice/HousePricePrediction/dataset.csv',index= False)
```

```python
# Fit the encoders on your training data
encoder_location = LabelEncoder()
encoder_condition = LabelEncoder()
encoder_garage = LabelEncoder()

# Assuming 'df' is your training data and columns are categorical
encoder_location.fit(df['Location'])
encoder_condition.fit(df['Condition'])
encoder_garage.fit(df['Garage'])

# Save the encoders
with open('encoder_location.pkl', 'wb') as f:
    pickle.dump(encoder_location, f)

with open('encoder_condition.pkl', 'wb') as f:
    pickle.dump(encoder_condition, f)

with open('encoder_garage.pkl', 'wb') as f:
    pickle.dump(encoder_garage, f)
```
